```python
import numpy as np
import pandas as pd

# Series를 생성해 보아요!
s = pd.Series([-1, 5, 10, 99],
              dtype=np.float64)
print(s)
print(s.values) # [-1.  5. 10. 99.]
print(s.index) # RangeIndex(start=0, stop=4, step=1)
print(s.dtype) # float64
```

```python
# Series는 생성할 때 index를 지정할 수 있어요!
# 지정인덱스를 사용할 수 있어요!
s = pd.Series([-1, 10, -8, 20],
              dtype=np.float64,
              index=['c', 'b', 'a', 'kkk'])
print(s)
print(s[1]) # 숫자 index를 이용해서 indexing => 10
print(s['a']) # -8
```

```python
# Series를 생성할 때 지정 index를 사용해서 Series를
# 생성할 꺼예요!
s = pd.Series([-1, 10, -8, 20],
              dtype=np.float64,
              index=[0, 2, 100, 6])
print(s)
print(s[2]) # 10 아하...지정index가 우선이군요!
# print(s[3]) # key error
```

```python
# Series를 생성할 때 지정 index를 사용해서 Series를
# 생성할 꺼예요!
# 같은 index를 사용할 수 있어요!
s = pd.Series([-1, 10, -8, 20],
              dtype=np.float64,
              index=['c', 'a', 'k', 'a'])
print(s)
print(s['a'])
```

```python
# Series에서 slicing은 어떻게 되나요?
s = pd.Series([-1, 10, -8, 20],
              dtype=np.float64,
              index=['c', 'a', 'k', 'b'])
print(s)
# print(s[0:2])
new_s = s[0:2]
# print(new_s)
new_s[0] = 100
print(new_s)
print(s)
```

```python
# Series에서 slicing은 어떻게 되나요?
s = pd.Series([-1, 10, -8, 20],
              dtype=np.float64,
              index=['c', 'a', 'k', 'b'])
print(s)
print(s['a':'b'])
```

```python
# Series에서 Fancy indexing할 수 있나요? => 가능
# Series에서 Boolean indexing을 할 수 있나요? => 가능
# Series도 Numpy의 ndarray에서 했던것처럼 집계함수가 있나요? => 네!
s = pd.Series([-1, 10, -8, 20],
              dtype=np.float64,
              index=['c', 'a', 'k', 'b'])
print(s[[0,2]])
# 현재 Series에서 -5보다 큰 값만 추출하세요!(Series형태로)
print(s[s > -5])
print(s.sum()) # 21.0
```

```python
# dictionary를 이용해서 Series를 만들어보아요!
import numpy as np
import pandas as pd

my_dict = {
    "서울": 3000,
    "인천": 2000,
    "제주": 1000
}

s = pd.Series(my_dict)
print(s)
print(s[0]) # 3000
print(s['제주']) # 1000
```

```python
# Series에 요소를 추가하거나 삭제하려면 어떻게 해야 하나요?
s = pd.Series([1,2,3,4])
print(s)
s[4] = 5
print(s)
# 값을 삭제할 경우에는 drop을 이용해요!
s = s.drop(2)
print(s)
```

```python
# A공장이 있어요!
# 2020-01-01부터 10일간의 생산량을 Series로 저장할꺼예요!
# 생산량은 평균이 50이고 표준편차가 5인 정규분포에서 랜덤하게 추출(정수)

# B공장이 있어요!
# 2020-01-05부터 10일간의 생산량을 Series로 저장할꺼예요!
# 생산량은 평균이 70이고 표준편차가 8인 정규분포에서 랜덤하게 추출(정수)

# 날짜별로 생산량의 합을 구할꺼예요!

import numpy as np
import pandas as pd
from datetime import date, datetime, timedelta
from dateutil.parser import parse

start_day = datetime(2020,1,1) # 2020-01-01 날짜에 대한 날짜값.
# 10일간의 생산량을 랜덤하게 생성해서 list를 만들어보아요!
a_factory_amount = [int(x) for x in np.random.normal(50,5,(10,))]
a_factory_index = [start_day + timedelta(days=x) for x in range(10)]
a_factory_s = pd.Series(a_factory_amount,
                        index=a_factory_index)
print(a_factory_s)

start_day = datetime(2020,1,5) # 2020-01-05 날짜에 대한 날짜값.
# 10일간의 생산량을 랜덤하게 생성해서 list를 만들어보아요!
b_factory_amount = [int(x) for x in np.random.normal(70,8,(10,))]
b_factory_index = [start_day + timedelta(days=x) for x in range(10)]
b_factory_s = pd.Series(b_factory_amount,
                        index=b_factory_index)
print(b_factory_s)

# 전체 생산량을 알기 위해서 두 공장의 생산량을 날짜별로
# 더해줄꺼예요!
print(a_factory_s + b_factory_s)
```

```python
# 이 문제를 해결해보세요!
# A 공장의 1월 11일~14일까지 지정인덱스를 추가하고 생산량을 0으로 설정
# B 공장의 1월 1일~4일까지 지정인덱스를 추가하고 생산량을 0으로 설정

index_a = set(a_factory_index)
print(index_a)
index_b = set(b_factory_index)
print(index_b)

index_a_b = index_a - index_b  # 1월1일 ~ 4일까지의 값이 남아요!
index_b_a = index_b - index_a  # 1월 11일 ~ 14일까지의 값이 남아요!

for i in index_a_b:
    b_factory_s[i] = 0

for i in index_b_a:
    a_factory_s[i] = 0

print(a_factory_s + b_factory_s)
```

```python
# DataFrame에 대해서 알아보아요!

# 먼저 DataFrame을 생성해보아요 => dictionary로 생성.

import numpy as np
import pandas as pd

data = {
    'name': ['홍길동','신사임당','강감찬', '이순신'],
    'year': [2020, 2021, 2020, 2023],
    'grade': [3.1, 4.0, 1.2, 2.5]
}
# Series를 만들때는 key값이 지정 index로 사용.
# DataFrame을 만들때는 key값이 컬럼명으로 사용.

# s = pd.Series(data)
df = pd.DataFrame(data)
display(df)

# 기억해야 하는 속성(기본속성)
print(df.shape) # (4, 3) => 총 데이터 12개
print(df.size) # 12
print(df.ndim) # 2
print(df.values)
```

```python
# DataFrame의 index, column, values

import numpy as np
import pandas as pd

data = {
    'name': ['홍길동','신사임당','강감찬', '이순신'],
    'year': [2020, 2021, 2020, 2023],
    'grade': [3.1, 4.0, 1.2, 2.5]
}

df = pd.DataFrame(data)
display(df)

print(df.index)  # RangeIndex(start=0, stop=4, step=1)
print(df.values) # 2차원 ndarray로 데이터만 추출
print(df.columns) # Index(['name', 'year', 'grade'], dtype='object')
print(df.columns[0]) # name
```

```python
# DataFrame을 생성한 후
# index와 컬럼명을 변경하고 싶어요!
data = {
    'name': ['홍길동','신사임당','강감찬', '이순신'],
    'year': [2020, 2021, 2020, 2023],
    'grade': [3.1, 4.0, 1.2, 2.5]
}

df = pd.DataFrame(data)
display(df)

new_df = df.rename(columns={'year':'입학연도',
                            'grade': '성적'},
                   index={0: 'zero'},
                   inplace=False)
display(new_df)
```
