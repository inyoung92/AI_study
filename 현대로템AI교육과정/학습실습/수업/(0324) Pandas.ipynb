```python
# Binning 처리(구간분할 처리)
# MPG Data set을 예로 처리해 보아요!

import numpy as np
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/KDTHome/[data]/auto-mpg.csv',
                 header=None)
df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower',
              'weight', 'acceleratioin', 'model_year', 'origin', 'name']
# display(df)
# df.info()
# print(df['horsepower'].value_counts())
# 해당 Series에 숫자가 아닌 값이 있는지 확인해 보아요!
# 현재 데이터 타입이 object로 되어 있어서 정규식을 이용해서 처리해야할듯
# 보여요!

# 사실 horsepower에 '?' 문자가 들어가 있어요!
# horserpower를 알 수 없어서 '?' 문자를 대신 데이터에 넣어놓은건대
# 이건 어떻게 처리해야 하나요?
df['horsepower'].replace('?', np.nan, inplace=True)
df.dropna(subset=['horsepower'], axis=0, inplace=True)
df['horsepower'] = df['horsepower'].astype('float')

# 이제 이 자동차의 마력(출력)을 Binning처리 해 보아요!
# 저출력, 보통 출력, 고출력의 형태로 Binning처리를 해 보아요!
# 이런 Binning처리할 때 일단 np가 가지고 있는 histogram()을
# 이용하면 편해요. 특정 구간에 데이터가 몇개 있는지 알려줘요!

count, divider =  np.histogram(df['horsepower'], bins=3)
print(count)  # [257 103  32]
print(divider) # [ 46.         107.33333333 168.66666667 230.        ]

bin_name = ['저출력', '보통출력', '고출력']

df['hp_cut'] = pd.cut(x=df['horsepower'],
                      bins=divider,
                      labels=bin_name,
                      include_lowest=True)

# display(df)

# 이렇게 category 데이터 형태로 기존 데이터를 변형 시킬 수 있어요!
# 여기서 하나 더 추가적으로 작업해야 하는 경우가 있어요!
horsepowser_dummies = pd.get_dummies(df['hp_cut'])
display(horsepowser_dummies)
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/KDTHome/[data]/auto-mpg.csv',
                 header=None)
df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower',
              'weight', 'acceleratioin', 'model_year', 'origin', 'name']

df['horsepower'].replace('?', np.nan, inplace=True)
df.dropna(subset=['horsepower'], axis=0, inplace=True)
df['horsepower'] = df['horsepower'].astype('float')

display(df['horsepower'])
# 해당 컬럼을 min-max scaling을 이용해서 정규화 시킬꺼예요!
max_min = df['horsepower'].max() - df['horsepower'].min()
df['horsepower_norm'] = ((df['horsepower'] - df['horsepower'].min())) / max_min
display(df)
```

```python
# DataFrame을 결합할 수 있어요!
# merge()를 이용해서 결합해요! => Database에서 했었던 join

data1 = {
    '학번': [1,2,3,4],
    '이름': ['홍길동','신사임당', '강감찬', '이순신'],
    '학년': [1,3,2,4]
}

data2 = {
    '학번' : [1,2,4,5],
    '학과' : ['철학', '국어국문', '기계', '컴퓨터'],
    '학점' : [3.1, 2.5, 1.6, 4.5]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

display(df1)
display(df2)

# inner join
# result = pd.merge(df1, df2, on='학번', how='inner')
# display(result)

# result = pd.merge(df1, df2, on='학번', how='left')
# result = pd.merge(df1, df2, on='학번', how='right')
result = pd.merge(df1, df2, on='학번', how='outer')
display(result)
```

```python
data1 = {
    '학번': [1,2,3,4],
    '이름': ['홍길동','신사임당', '강감찬', '이순신'],
    '학년': [1,3,2,4]
}

data2 = {
    '학생학번' : [1,2,4,5],
    '학과' : ['철학', '국어국문', '기계', '컴퓨터'],
    '학점' : [3.1, 2.5, 1.6, 4.5]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

# 컬럼명이 다른 경우에 merge해야 하는 경우
result = pd.merge(df1, df2, left_on='학번', right_on='학생학번',
                  how='inner')
display(result)
```

```python
# Raw data Loading
titanic = sns.load_dataset('titanic')
# display(titanic)
# print(titanic.shape)
# 모든 행에 대해 age, fare컬럼만 들고 올꺼예요!
df = titanic.loc[:,['age', 'fare']]
display(df.head())

# 정규화(Normalization) - 최대,최소를 이용한 정규화
# MinMaxScaling을 구현해 보아요!

# 사용자 정의 함수
def min_max(s):
    return (s - s.min()) / (s.max() - s.min())

result = df.apply(min_max, axis=0)
display(result)
```

```python
# DataFrame의 Grouping

import numpy as np
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset('titanic')
# 필요한 몇개의 컬럼만 추려서 사용할꺼예요!

df = titanic.loc[:,['age', 'sex', 'class', 'fare', 'survived']]
# display(df.head())

# 그룹을 조금 복잡하게 설정할께요!
# 그룹안에 다시 그룹을 만들꺼예요!
grouped = df.groupby(['class', 'sex'])

# for key, group in grouped:
#     print(key)
#     display(group.head(3))

result = grouped.mean()
display(result)

group3 = grouped.get_group(('Third','female'))
display(group3)
```

```python
titanic = sns.load_dataset('titanic')
# 필요한 몇개의 컬럼만 추려서 사용할꺼예요!

df = titanic.loc[:,['age', 'sex', 'class', 'fare', 'survived']]
# display(df.head())

grouped = df.groupby('class')
display(grouped.survived.mean())

# age열의 평균값이 30보다 작은 그룹만을 따로 선택.
# 생성된 그룹에 대해 조건을 설정해 조건을 만족하는 그룹만
# 따로 선택하려면 어떻게 해야 하나요
# filter()

group_filter = grouped.filter(lambda x:x.age.mean() < 30)
display(group_filter.tail())
```
