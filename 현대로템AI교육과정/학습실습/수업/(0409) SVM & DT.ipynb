```python
# Iris Data Set을 이용해서
# SVM과 Regression(Logistic)을 이용해서 분류모델을
# 만들고 성능을 비교해 보아요!
# 필요한 module import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
iris = load_iris()
df = pd.DataFrame(iris['data'],
                  columns=iris['feature_names'])
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
df['label'] = iris.target
display(df)
```

```python
# feature engineering & 데이터 전처리
# 특정 feature(독립변수)를 제거하는경우
# (상관관계가 없거나 NaN이 너무 많거나 다른 Feature와 중복되거나)
# 데이터 전처리
# 1. 결측치가 없어요.
# 2. 이상치도 특별히 없어요!
# 2.1. 중복데이터 처리
# 2.2. 불균형 데이터 처리
# 3. 정규화
x_data = df.drop('label', axis=1, inplace=False).values
t_data = df['label'].values
scaler = MinMaxScaler()
scaler.fit(x_data)
x_data_norm = scaler.transform(x_data)
# 4. 학습데이터와 테스트 데이터분리
x_data_train_norm, x_data_test_norm, t_data_train, t_data_test = \
train_test_split(x_data_norm,
                 t_data,
                 test_size=0.3,
                 stratify=t_data)
```

```python
# Model 구현
logistic_model = LogisticRegression()
logistic_model.fit(x_data_train_norm,
                   t_data_train)
print(classification_report(t_data_test,
                            logistic_model.predict(x_data_test_norm)))
```

```python
svm_model = SVC(C=10, kernel='rbf')
svm_model.fit(x_data_train_norm,
              t_data_train)
print(classification_report(t_data_test,
                            svm_model.predict(x_data_test_norm)))
```

```python
# 좋은 Hyperparameter를 찾기위해
# GridSearchCV를 이용해볼꺼예요!
from sklearn.model_selection import GridSearchCV

param_grid = [{
    'kernel': ['linear'],
    'C': [0.1, 0.5, 1, 10, 100]
},{
    'kernel': ['rbf'],
    'C': [0.1, 0.05, 0.7, 1, 5, 10, 100],
    'gamma' : [0.001, 0.01, 0.1, 1]
}]

model = SVC()

grid_search = GridSearchCV(model,
                           param_grid,
                           cv=5,
                           scoring='accuracy',
                           verbose=1)

grid_search.fit(x_data_train_norm,
                t_data_train)
```

```python
# 가장 좋은 parameter는 어떤것인고
# 그때 score는 얼마인가요?
print(f'가장 좋은 param : {grid_search.best_params_}')
print(f'그때의 score : {grid_search.best_score_}')
```

```python
%reset
# 이번에는 Decision Tree를 이용해서
# Iris Data set에 대한 분류모델을 만들어 보아요!
# 꽃받침의 길이와 너비, 꽃잎의 길이와 너비 총 4개의 독립변수를
# 이용. 종속변수 3분류의 Iris 품종이 나와요
# 독립변수를 2개만 이용해서 2차원 평면에 결정경계를 그려볼꺼예요!
# Decision Tree는 일반적으로 연속적인 숫자가 아닌 discrete한
# value를 사용하기 때문에 정규화를 하지 않아도 크게 문제가 되지
# 않아요!
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from mlxtend.plotting import plot_decision_regions
from sklearn.metrics import classification_report
# 1. Regression
# 2. SVM
# 3. Decision Tree
# 4. KNN
```

```python
# 데이터 로딩 및 전처리
iris = load_iris()
df = pd.DataFrame(iris['data'],
                  columns=iris['feature_names'])
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
df['label'] = iris.target
# display(df)
df = df.drop_duplicates()
# 1. 결측치 이상치 없구요!
# 2. 정규화를 위해서 데이터를 일단 x_data, t_data로 분리
x_data = df[['sepal_length', 'petal_length']].values
t_data = df['label'].values
scaler = MinMaxScaler()
scaler.fit(x_data)
x_data_norm = scaler.transform(x_data)
x_data_train_norm, x_data_test_norm, t_data_train, t_data_test = \
train_test_split(x_data_norm,
                 t_data,
                 test_size=0.3,
                 stratify=t_data)
```

```python
# DataSet은 MNIST Data Set을 사용할꺼예요!
df = pd.read_csv('/content/train.csv')
# display(df)
# 1. 결측치 이상치 없구요!
# 2. 정규화를 위해서 데이터를 일단 x_data, t_data로 분리
x_data = df.drop('label', axis=1, inplace=False).values
t_data = df['label'].values
scaler = MinMaxScaler()
scaler.fit(x_data)
x_data_norm = scaler.transform(x_data)
x_data_train_norm, x_data_test_norm, t_data_train, t_data_test = \
train_test_split(x_data_norm,
                 t_data,
                 test_size=0.3,
                 stratify=t_data)
```
