```python
# 제공된 excel파일을 이용해서
# DataFrame을 생성하고 적절한 line plot을 그려보아요!

# excel 파일을 불러들여서 처리해야 해요!
# excel 파일을 로딩하기 위해서는 openpyxl 모듈이 필요해요!
# 이미 설치되어 있을거 같아요!
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('./lineplot_sample_data.xlsx')

# display(df.head(30))
# DataFrame을 살펴봤더니.. 전출지별 column에 NaN(결측치)가 포함되어 있어요.
# 변환과정에서 약간 문제가 있어서 값이 NaN으로 설정되었는데 바로 앞의 행의 값으로
# NaN값을 변경시켜 주면 될 거 같아요!
# df = df.fillna(method='ffill')
df = df.ffill()

display(df.head(30))
```

```python
# 우리가 관심있는 정보는
# 서울특별시에서 다른 지역으로 이동한 데이터.
# 전출지별 컬럼에서 서울특별시 라는 값을 갖는 데이터만 추출
# 전입지별 컬럼에서 서울특별시 값을 제외시킬꺼예요!

mask = (df['전출지별'] == '서울특별시') & (df['전입지별'] != '서울특별시')
```

```python
df_seoul = df[mask]
# display(df_seoul)
# 추가적으로 몇가지 작업만 더 해봐요!
# '전출지별' 컬럼은 필요가 없으니까 삭제할꺼구요!
df_seoul = df_seoul.drop('전출지별', axis=1, inplace=False)
# display(df_seoul)
# '전입지별' 컬럼의 이름은 '전입지'로 변경할꺼예요!
df_seoul = df_seoul.rename({'전입지별' : '전입지'}, axis=1, inplace=False)
# display(df_seoul)
# 기본적으로 숫자 인덱스로 되어 있는데 '전입지' 컬럼을 지정 index로 설정할꺼예요!
df_seoul = df_seoul.set_index('전입지', inplace=False)
# display(df_seoul)

# 이제 선 그래프를 그리기 위해서
# 경기도만 추출할 꺼예요!(서울에서 경기도로 전입한 사람의 연도별 추이)
s = df_seoul.loc['경기도']
print(s)
```

```python
# 서울에서 경기도로 전입한 사람들의 연도별 수를 알아냈어요(Series)
# 이 데이터를 이용해서 line plot을 그려보아요!
plt.plot(s.index, s.values)
plt.show()
```

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from matplotlib import rc
import numpy as np
import pandas as pd

# 일부 한글 폰트를 사용할 경우 마이너스 부호가 표현될 때 문제가 되는 경우가 있어요!
# 마이너스 부호('-') 대신에 대쉬 부호('-')를 사용하면 문제가 해결되요!
mpl.rcParams['axes.unicode_minus'] = False

# 한글 폰트 처리를 해야지 한글이 정상적으로 출력되요!
font_path = './malgun.ttf'
font_name = fm.FontProperties(fname=font_path).get_name()  # 폰트이름을 알아와요!
# print(font_name)  # Malgun Gothic
rc('font', family=font_name)

# 그래프의 스타일을 지정할 수 도 있어요!
plt.style.use('ggplot')
# 그래프의 그림 크기를 지정할 수 있어요!
plt.figure(figsize=(10,5))  # 그림의 가로길이 세로길이를 tuple로 명시(단위는 인치)
# x축 눈금(연도)가 겹쳐서 알아보기 어려워요!
# label을 회전시켜서 겹쳐보이지 않게 처리
plt.xticks(rotation='vertical')

plt.title('서울 -> 경기 인구 이동')
plt.ylabel('인구 이동수')
plt.xlabel('연도')


# 그래프를 그릴때 약간의 속성들을 줄 수 있어요!
# marker나 색상, 추가적인 문자열을 사용할 수 있어요!

plt.plot(s.index, s.values,
        marker = 'o',
        markersize=10,
        markerfacecolor='y',
        color='b',
        linewidth=3)

# 범례
plt.legend(labels=['서울 -> 경기'], loc='best')

# y축 범위를 증가시켰어요. 주석을 포함할 공간을 확보할 목적으로..
plt.ylim(5000, 800000)

# 화살표를 표시
plt.annotate("",  # 화살표 시작부분에 문자열을 추가가
            xy=(20,620000),  # 화살표의 끝점. x좌표는 index, y좌표는 값을 이용
            xytext=(2,290000), # 화살표의 시작점.
            xycoords='data',
            arrowprops=dict(arrowstyle='->',
                           color='skyblue',
                           lw=5)
            )
# 텍스트를 표시
plt.annotate("인구 이동 증가(1970-1995)",  # 화살표 시작부분에 문자열을 추가가
             xy=(10,350000),  # 텍스트 위치 기준점
             rotation=30,
             va='baseline',
             ha='center',
             fontsize=15
            )
plt.annotate("",  # 화살표 시작부분에 문자열을 추가가
            xy=(47,450000),  # 화살표의 끝점. x좌표는 index, y좌표는 값을 이용
            xytext=(30,600000), # 화살표의 시작점.
            xycoords='data',
            arrowprops=dict(arrowstyle='->',
                           color='olive',
                           lw=5)
            )
# 텍스트를 표시
plt.annotate("인구 이동 감소(1995-2017)",  # 화살표 시작부분에 문자열을 추가가
             xy=(40,500000),  # 텍스트 위치 기준점
             rotation=-15,
             va='baseline',
             ha='center',
             fontsize=15
            )
plt.show()
```
