```python
import os

# 이미지와 라벨 폴더 경로
image_folder = '/home/govlept1004/jupyter_home/tank_project/od_image'   # 이미지 폴더 경로 수정
label_folder = '/home/govlept1004/jupyter_home/tank_project/od_txt'   # 라벨 폴더 경로 수정

# PNG 파일 목록 정렬
image_files = [f for f in os.listdir(image_folder) if f.endswith('.png')]
image_files.sort()

for i, image_file in enumerate(image_files, start=1):
    base_old = os.path.splitext(image_file)[0]  # 'capture_0000'
    old_image_path = os.path.join(image_folder, image_file)
    old_label_path = os.path.join(label_folder, base_old + '.txt')

    # 새 이름 구성
    new_base = f'od_img_{i}'
    new_image_path = os.path.join(image_folder, new_base + '.png')
    new_label_path = os.path.join(label_folder, new_base + '.txt')

    # 파일 존재 확인 후 이름 변경
    if os.path.exists(old_label_path):
        os.rename(old_image_path, new_image_path)
        os.rename(old_label_path, new_label_path)
    else:
        print(f"[주의] 라벨 파일 없음: {old_label_path} → 해당 쌍은 건너뜀")

print("✅ 이미지 및 라벨 이름 일괄 변경 완료!")
```

```python
import shutil

# 원본 폴더 경로 (JupyterLab에서 저장된 위치)
source = '/home/govlept1004/jupyter_home/tank_project/od_image'  # 예: './output' 또는 'data/result'

# 바탕화면 경로 (Windows 예시)
desktop = os.path.join(os.path.expanduser("~"), "Desktop")

# 이동할 대상 경로
destination = os.path.join(desktop, 'od_img')  # 원하는 이름 지정

# 폴더 이동
shutil.move(source, destination)

print("✅ 바탕화면으로 이동 완료!")
```
