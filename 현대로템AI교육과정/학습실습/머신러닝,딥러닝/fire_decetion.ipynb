```python
def clean_manifest_file(file_path):
    with open(file_path, 'r') as f:
        lines = f.readlines()

    cleaned_lines = []
    for line in lines:
        # 파일명에서 공백과 괄호 제거하고 언더스코어로 변경
        new_line = line.replace(" (", "_").replace(")","").replace(" ", "_")
        cleaned_lines.append(new_line)

    with open(file_path, 'w') as f:
        f.writelines(cleaned_lines)

# 절대 경로로 지정
train_manifest = "/home/govlept1004/jupyter_home/YOLO_Object_Detection/darknet/data/fire_train_manifest.txt"
val_manifest = "/home/govlept1004/jupyter_home/YOLO_Object_Detection/darknet/data/fire_val_manifest.txt"

clean_manifest_file(train_manifest)
clean_manifest_file(val_manifest)

print("파일 경로가 수정되었습니다.")
```
