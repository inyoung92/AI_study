1. class이름이 명시된 파일을 하나 생성해야되요!
   .pbtxt라고 해요! 나름대로의 형식이 있어요~ 코드로 만들어보아요!(해당 코드는 노션에 명시!)

```python
voc_classes = [
    "fire"
]

with open("fire_label_map.pbtxt", "w") as f:
    for idx, class_name in enumerate(voc_classes, 1):
        f.write("item {\n")
        f.write(f"  id: {idx}\n")
        f.write(f"  name: '{class_name}'\n")
        f.write("}\n\n")
```
