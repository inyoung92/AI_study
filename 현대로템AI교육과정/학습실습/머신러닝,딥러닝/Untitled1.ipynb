```python
import os

# 경로 설정 (파일들이 있는 경로로 수정)
folder_path = '/home/govlept1004/jupyter_home/YOLO_Object_Detection/fire_VOC/fire-dataset/fire-dataset/train/annotations'  # 정확한 경로 입력

# 폴더 내 파일들 확인
files = os.listdir(folder_path)

# 파일명 수정
for filename in files:
    # 공백과 괄호 제거하고, 숫자는 밑줄(_)로 연결
    new_filename = filename.replace(' ', '_').replace('(', '').replace(')', '')
    
    # 파일 이름이 바뀌었으면 이름 변경
    if new_filename != filename:
        os.rename(os.path.join(folder_path, filename), os.path.join(folder_path, new_filename))

# 수정된 파일들 확인
new_files = os.listdir(folder_path)
print(new_files[:10])  # 처음 10개 파일명 확인
```

```python
def clean_manifest_file(file_path):
    with open(file_path, 'r') as f:
        lines = f.readlines()

    cleaned_lines = []
    for line in lines:
        # 파일명에서 공백과 괄호 제거하고 언더스코어로 변경
        new_line = line.replace(" (", "_").replace(")", "").replace(" ", "_")
        cleaned_lines.append(new_line)

    with open(file_path, 'w') as f:
        f.writelines(cleaned_lines)

# 수정할 manifest 파일 경로 지정
train_manifest = "/home/govlept1004/jupyter_home/YOLO_Object_Detection/darknet/data/fire_train_manifest.txt"


# manifest 파일 정리
clean_manifest_file(train_manifest)

print("파일 경로가 수정되었습니다.")
```
