```python
# Attention 연산의 간단 버전
# 입력 문장
# -> 쿼리, 키, 벨류 데이터를 생성
# -> 쿼리와 키의 유사도를 계산(내적) -> Attention score
# -> Attention score를 이용해서 value의 가중 합 -> Attention Output
```

```python
import numpy as np
# 입력 데이터는 "나는 너를 사랑해"
# 입력 데이터는 => [”나는”, ”너를”, ”사랑해”]
token = ["나는", "너를", "사랑해"]
# "나는" => [1.0, 0.0, 1.0] (embedding)
# "너를" => [1.0, 1.0, 1.0] (embedding)
# "사랑해" => [1.0, 1.0, 0.0] (embedding)

# 원래는 값을 변환시켜서 사용해야 해요!
# 근데 계산이 힘들어서 같은 값을 이용할게요

key_vector = np.array([
    [1.0, 0.0, 1.0], # 나는
    [1.0, 1.0, 1.0], # 너를
    [1.0, 1.0, 0.0] # 사랑해
])

value_vector = np.array([
    [10, 0, 0], # 나는
    [0, 10, 0], # 너를
    [0, 0, 10] # 사랑해
])

query = key_vector[2]  #  [1.0, 1.0, 0.0] # 사랑해

# 1. 유사도 계산(내적을 이용해서 계산)
score = np.dot(key_vector, query)
# [1, 2, 2]
print(score)
```

```python
# 2. 숫자의 안정화를 시키기 위해서 embedding의 차원의 root값을 이용해 값을 나눠줘야 해요
# 원래는 np.sqrt(3)를 구해서 각각의 값을 나눠줘야 해요
# 지금은 생략!

# 3. softmax를 이용해서 확률값으로 변경
# 확률값으로 변경하려면 사실 쉬운방법이 있어요!
# 몽땅 더한 후 그 값으로 각각의 값을 나눠주면 되요!
# [0.2, 0.4, 0.4] -> 값이 0이 나올 수 있기 때문에 이렇게 단순한 식은 사용할 수 없다
# 우리가 원하는 것은 값이 쿤거에 더 가중치를 주어서 확률값으로 변경하는 거에요
def softmax(x):  # x -> [1. 2. 2.]
    e_x = np.exp(x - np.max(x))  
    # 지수함수라서 값이 커지면 급격하게 결과값이 커지기 시작한다. 
    # 그래서 안정화 시키기위해 (x - np.max(x)) 이렇게 최대값을 빼준다. 
    return e_x / e_x.sum()

attention_weighted = softmax(score)
print(attention_weighted) # [0.1553624 0.4223188 0.4223188] (3,)
# 결국 가중치를 구했어요~
print(attention_weighted[:, np.newaxis])  # (3,1)
# 이를 이용해서 Attention Output을 구할 수 있어요!
print(np.sum(attention_weighted[:, np.newaxis] * value_vector, axis=0))  # 컬럼별로(행별로) 더해라
# [1.55362403 4.22318798 4.22318798]
# 이 값은 "사랑해"라는 Query의 Attention Output
```
